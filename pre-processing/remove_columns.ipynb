```python
import math
import asyncio.taskgroups
import asyncio.taskgroups
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

AAPL = pd.read_csv("../processed_data/AAPL_combined.csv")
AMZN = pd.read_csv("../processed_data/AMZN_combined.csv")
FB = pd.read_csv("../processed_data/FB_combined.csv")
NVDA = pd.read_csv("../processed_data/NVDA_combined.csv")
TSLA = pd.read_csv("../processed_data/TSLA_combined.csv")

ls = [AAPL, AMZN, FB, NVDA, TSLA]

cols_to_drop = ['links', 'reshares', 'reshare_message', 'structurable, mentioned_users, conversation']
for i in ls:
    for col in cols_to_drop:
        if col in i.columns:
            del i[col]

AAPL.to_csv("AAPL_clean.csv",  index=False)
AMZN.to_csv("AMZN_clean.csv",  index=False)
FB.to_csv("FB_clean.csv",  index=False)
NVDA.to_csv("NVDA_clean.csv",  index=False)
TSLA.to_csv("TSLA_clean.csv",  index=False)
```
