```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

AAPL = pd.read_csv("AAPL_clean.csv")
AMZN = pd.read_csv("AMZN_clean.csv")
FB = pd.read_csv("FB_clean.csv")
NVDA = pd.read_csv("NVDA_clean.csv")
TSLA = pd.read_csv("TSLA_clean.csv")


ls = [AAPL, AMZN, FB, NVDA, TSLA]

for df in ls:
    null_counts = df.isnull().sum()
    
    # Filter to show only columns with > 0 nulls (optional)
    if null_counts.sum() == 0:
        print("  No null values found.")
    else:
        print(null_counts[null_counts > 0])
        
    # percentage of missing data
    print((df.isnull().sum() / len(df)) * 100)
```
